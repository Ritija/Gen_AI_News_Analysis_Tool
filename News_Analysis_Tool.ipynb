```python
import os
import json
import pandas as pd
from openai import OpenAI
from dotenv import load_dotenv

# Load API key from .env
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")

if api_key:
    print("✅ API key loaded successfully.")
else:
    print("❌ API key not found. Please check your .env file.")

client = OpenAI(api_key=api_key)
```

```python
def get_completion(user_message, system_message=None, model="gpt-4o-mini", json_mode=False):
    messages = []
    if system_message:
        messages.append({"role": "system", "content": system_message})
    messages.append({"role": "user", "content": user_message})

    if json_mode:
        response = client.chat.completions.create(
            model=model,
            messages=messages,
            response_format={"type": "json_object"}
        )
    else:
        response = client.chat.completions.create(
            model=model,
            messages=messages
        )

    content = response.choices[0].message.content
    return content.strip() if content else None
```

```python
sentiments = [] 
for article in articles:
    
    summary = get_completion(
        article,
        system_message="Summarise the following text in exactly one sentence. Be concise and preserve the key idea.Summary should be in less than 10 words"
    )
    
    result_classification = json.loads(get_completion(
        article,
        system_message=f"Classify the news article into exactly one of these categories: {CATEGORIES}. Reply in JSON: {{category: string, confidence: high|medium|low}}",
        json_mode=True
    ))

    
    result_sentiment = json.loads(get_completion(
        article,
        system_message="You are a sentiment analyser. Reply in JSON: {sentiment: POSITIVE|NEGATTIVE|NEUTRAL}",
        json_mode=True
    ))
    sentiments.append(result_sentiment)

    

    

   # print(f"Original Article : {article}")
    print(f"Summary  : {summary}")
    print(f"Sentiment: {result_sentiment['sentiment'].upper()}")
    print(f"Category  : {result_classification['category']}  ({result_classification['confidence']} confidence)")
    print("--------------------------------------------------------------------------")
```
